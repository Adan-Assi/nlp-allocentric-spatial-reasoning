```python
import pandas as pd

df = pd.read_parquet(
    "/vol/joberant_nobck/data/NLP_368307701_2526a/adanassi/"
    "nlp-allocentric-spatial-reasoning/reports/llm_audits/"
    "LLM_DEGRADATION_INPUT.parquet"
)

masked_answerable = df[
    (df['variant_type'] != 'original') &
    (df['oracle_label'] == 'Answerable')
][['sample_id', 'city', 'variant_type', 'oracle_label', 'masked_instruction']].copy()

print(f"Total masked+Answerable: {len(masked_answerable)}")
print(f"\nBy city:\n{masked_answerable['city'].value_counts().to_string()}")
print(f"\nBy variant type:\n{masked_answerable['variant_type'].value_counts().to_string()}")
print(f"\nSample examples:")
for city in masked_answerable['city'].unique():
    subset = masked_answerable[masked_answerable['city'] == city]
    row = subset.iloc[0]
    print(f"\n--- {city} | {row['variant_type']} ---")
    print(row['masked_instruction'])
```
